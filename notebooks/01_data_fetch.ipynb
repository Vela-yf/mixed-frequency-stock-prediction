```python
import pandas as pd
import time
import numpy as np
from WindPy import w
```

```python
w.start() # type: ignore
```

```python
stock_codes_300 = pd.read_csv('300.csv', usecols=['wind_code'])["wind_code"].tolist()
print(stock_codes_300)
stock_codes_500 = pd.read_csv('500.csv', usecols=['wind_code'])["wind_code"].tolist()
print(stock_codes_500)
```

```python
result = w.wsd("600000.SH", "open,high,low,close,volume,amt", "2016-01-01", "2026-02-27", "unit=1;Currency=CNY;PriceAdj=B")
df = pd.DataFrame(data = np.array(result.Data).T, 
                  columns=result.Fields, 
                  index=result.Times)
df["code"] = result.Codes[0]
df = df.reset_index().rename(columns={"index": "date"})
print(df.head(5))
```

```python
all_dfs = []
for i in range(len(stock_codes_300)):
    stock_code = stock_codes_300[i]
    result = w.wsd(stock_code, "open,high,low,close,volume,amt", "2016-01-01", "2026-02-27", "unit=1;Currency=CNY;PriceAdj=B")
    df = pd.DataFrame(data = np.array(result.Data).T, 
                      columns=result.Fields, 
                      index=result.Times)
    df["code"] = result.Codes[0]
    df = df.reset_index().rename(columns={"index": "date"})
    all_dfs.append(df)
    time.sleep(0.1)
    print(f"第{i+1}只股票{stock_code}数据已获取")
final_df = pd.concat(all_dfs, axis=0, ignore_index=True)
```

```python
print(len(final_df))
final_df.to_parquet("300_data.parquet", index=False)
```

```python
# 对财务数据
def get_financial_data(stock_code):
    result = w.wsd(stock_code, "west_netprofit_fy1_chg_3m,west_netprofit_ftm_chg_3m,west_sales_fy1_3m,west_sales_ftm_chg_3m,val_avgpe_sw,pe_ttm,val_stdpe_sw,dfa_tltshare,fa_roe_avg,fa_debttoasset,fa_orgr_ttm,fa_ctop", "2016-01-01", "2026-02-27", "unit=1;Currency=CNY;PriceAdj=B")
    df = pd.DataFrame(data = np.array(result.Data).T, 
                      columns=result.Fields, 
                      index=result.Times)
    df["code"] = result.Codes[0]
    df = df.reset_index().rename(columns={"index": "date"})
    return df
```

```python
# 对沪深300
all_financial_dfs_300 = []
for i in range(len(stock_codes_300)):
    stock_code = stock_codes_300[i]
    df = get_financial_data(stock_code)
    all_financial_dfs_300.append(df)
    time.sleep(0.1)
    print(f"第{i+1}只股票{stock_code}财务数据已获取")
final_financial_df_300 = pd.concat(all_financial_dfs_300, axis=0, ignore_index=True)
print(len(final_financial_df_300))
final_financial_df_300.to_parquet("300_financial_data.parquet", index=False)
```

```python
# 对中证500
all_financial_dfs_500 = []
for i in range(len(stock_codes_500)):
    stock_code = stock_codes_500[i]
    df = get_financial_data(stock_code)
    all_financial_dfs_500.append(df)
    time.sleep(0.1)
    print(f"第{i+1}只股票{stock_code}财务数据已获取")
final_financial_df_500 = pd.concat(all_financial_dfs_500, axis=0, ignore_index=True)
print(len(final_financial_df_500))
final_financial_df_500.to_parquet("500_financial_data.parquet", index=False)
```
